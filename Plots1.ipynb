```python
2. Box Plot (Box-and-Whisker Plot)
---------------------------------------

This is the second most essential plot every data analyst must master

A box plot visualizes distribution + outliers in a single chart.

✔ It shows:

Median (Q2)
25th percentile (Q1)
75th percentile (Q3)
IQR (Inter-Quartile Range = Q3 − Q1)
Minimum & Maximum values (without outliers)
Outliers (points outside 1.5 × IQR)

The box is longer upward
------------------------
This means the upper salaries have more spread / variation.

✔ The median is closer to the bottom of the box                        

This means:
➡ Right-skewed distribution
(i.e., a few people earn more)

✔ The top whisker is longer

High-earners stretch the upper range.

✔ Lower whisker is short

Low salaries are closer together.
```

```python
# violin plot
--------------
-> It is the the combination of by default KDE + Box plot

4. Violin Plot

This is one of the cleanest and most powerful univariate plots because it combines:

✔ KDE Plot (shape of distribution)
✔ Box Plot (median, quartiles)

So a violin plot = KDE + Boxplot in one visual.

This makes it extremely useful in HR analytics, finance analytics, clinical data, and machine learning EDA.

A violin plot shows ALL of these in one:
---------------------------------------

✔ Distribution shape (like KDE)
✔ Symmetry or skewness
✔ Peaks (unimodal or multimodal)
✔ Spread (width)
✔ Quartiles (from the mini-boxplot inside)
✔ Density of values
✔ Whether the data has long tails

No other plot gives this much information at once


Outer Shape Explained (Deep Understanding)
-------------------------------------------
Wider part of violin = high density = more employees in that salary range
Narrow part of violin = low density = fewer employees in that salary range

This outer shape is basically a mirrored KDE plot (left + right).

Think of it like a mountain:

The mountain is wide where many people earn similar salaries

The mountain becomes thin where only few people earn salaries

Density = smooth estimate of how crowded an area is
```

```python
#joint plot
#-----------

-> It is nothing but the combination of Instagram plot and scatter plot.
-> x,y axis -> histogram
-> middle part -> scatter plot


A joint plot shows:
---------------------
Relationship between two variables
AND their individual distributions at the same time


Normally, you need 3 separate plots:
----------------------------------

Scatter plot → relationship
Histogram of X → distribution
Histogram of Y → distribution

👉 Joint plot combines all 3 in one view


| Area   | What it shows              |
| ------ | -------------------------- |
| Center | Relationship between X & Y |
| Top    | Distribution of X          |
| Right  | Distribution of Y          |

kind : { "scatter" | "kde" | "hist" | "hex" | "reg" | "resid" }
    Kind of plot to draw. See the examples for references to the underlying functions
```
