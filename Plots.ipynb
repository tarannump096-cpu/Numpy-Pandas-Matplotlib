```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

x=np.random.normal(4,15,100)

sns.distplot(x)
plt.title(' Normal Distribution')
plt.xlabel('x values')
plt.ylabel(' frequency of x')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

x=np.random.exponential(4,100)
plt.title(' RightSkwed Distribution')
plt.xlabel('x values')
plt.ylabel(' frequency of x')
sns.distplot(x)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

x=100-np.random.exponential(4,100)
plt.title(' LeftSkiwed Distribution')
plt.xlabel('x values')
plt.ylabel(' frequency of x')
sns.distplot(x)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

x=np.random.uniform(4,50,100)
plt.title(' Uniform Distribution')
plt.xlabel('x values')
plt.ylabel(' frequency of x')
sns.distplot(x)
plt.show()
```

```python
## Box Plot
#-> the box plot is the main thing is to finds Outliers ,mean median,Q1(25% percentile),Q3(75%)

x=[4,5,10,20,15,32,27,28,69,80,250,290,500]

plt.boxplot(x)
plt.title(' Find outliers of X column')
plt.xlabel(' Values of X column')
plt.ylabel(' density of Y column')
plt.show()
```

```python
2. Box Plot (Box-and-Whisker Plot)
---------------------------------------

This is the second most essential plot every data analyst must master

A box plot visualizes distribution + outliers in a single chart.

✔ It shows:

Median (Q2)
25th percentile (Q1)
75th percentile (Q3)
IQR (Inter-Quartile Range = Q3 − Q1)
Minimum & Maximum values (without outliers)
Outliers (points outside 1.5 × IQR)

The box is longer upward
------------------------
This means the upper salaries have more spread / variation.

✔ The median is closer to the bottom of the box                        

This means:
➡ Right-skewed distribution
(i.e., a few people earn more)

✔ The top whisker is longer

High-earners stretch the upper range.

✔ Lower whisker is short

Low salaries are closer together.
```

```python
## Box Plot
#-> the box plot is the main thing is to finds Outliers ,mean median,Q1(25% percentile),Q3(75%)

x=[4,5,10,20,15,32,27,28,69,80,250,290,500]

plt.boxplot(x,whis=(2,5))
plt.title(' Find outliers of X column')
plt.xlabel(' Values of X column')
plt.ylabel(' density of Y column')
plt.show()
```

```python
# KDE ( kernel density estimates)
------------------------------------
-> It display smooth of the density.
-> It works on how the data points are conquered at a 1 place.
-> This is sea borne plot


3.KDE Plot (Kernel Density Estimation Plot)
--------------------------------------------

This is one of the MOST important plots in data analysis because it shows the smooth probability distribution of a numeric column.

It is basically a smoothed histogram → looks like a mountain curve.


Shape of the distribution

(normal, right-skewed, left-skewed)

✔ Where most values are concentrated

(peaks)

✔ Spread of the data

(width)

✔ Whether the data is multimodal

(e.g., 2 peaks → 2 groups)

✔ Better than histogram because:

No bin-size confusion

Smooth and easy to compare
```
